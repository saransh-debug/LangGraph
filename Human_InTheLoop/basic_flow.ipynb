```python
from langgraph.graph import StateGraph ,START , END 
from typing import TypedDict,Annotated
from langchain_core.messages import AIMessage ,HumanMessage,BaseMessage
from langgraph.graph.message import add_messages
from langgraph.checkpoint.memory import InMemorySaver
from langchain_groq import ChatGroq
from dotenv import load_dotenv
from langgraph.types import interrupt,Command
```

```python
llm = ChatGroq( model="openai/gpt-oss-20b",
    temperature=0.7)
```

```python
checkpointer = InMemorySaver()
config = {"configurable":{
    "thread_id":"thread-1"
}
    
}
```

```python
class graph_state(TypedDict):

    messages:Annotated[list[BaseMessage] , add_messages]
```

```python
def chat_node(state:graph_state):
    decision = interrupt(
        {"value":"approval",
         "reason":"The llm will answer the user's question based on the output",
         "question":state['messages'][-1].content,
         "instruction":"Is this question is worthy enough to be asked to llm?"
    })

    if decision['response']=="no" or decision['response']=="No":
        return {"messages":(AIMessage(content="not worthy enough , hence rejected."))}
    elif decision['response']=="yes" or decision['response']=="Yes":
        res = llm.invoke(state['messages'])
        return {'messages':[res]}
```

```python
graph = StateGraph(graph_state)

graph.add_node("chat",chat_node)

graph.add_edge(START ,"chat")
graph.add_edge("chat",END)
```

```python
compiler = graph.compile(checkpointer=checkpointer)
compiler
```

```python
res = compiler.invoke(
    {"messages":[HumanMessage(content="which is better ghost of tshushima or rdr2 , choose one ")]},
    config=config
)

res
```

```python
ai_question = res['__interrupt__'][0].value
ai_question
```

```python
user_input = input(f"Enter your input:{ai_question} ")
```

```python
final_res = compiler.invoke(
    Command(resume={"response":user_input}),
    config=config
)
print(final_res)
```
