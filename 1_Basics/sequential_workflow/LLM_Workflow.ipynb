```python
from langgraph.graph import StateGraph , START , END 
from langchain_google_genai import ChatGoogleGenerativeAI
from typing import TypedDict
from dotenv import load_dotenv

load_dotenv()
```

```python
model = ChatGoogleGenerativeAI(model='gemini-2.5-flash')
```

```python
class State_Schema(TypedDict):

    title:str
    outline:str 
    content:str  
    score:str

```

```python
graph = StateGraph(State_Schema)
```

```python
def gen_outline(State:State_Schema)->State_Schema:
    title = State['title']
    prompt = f"genrate an outline for the following topic:{title}"
    res = model.invoke(prompt)
    State['outline'] = res.content
    return State

def gen_blog(State:State_Schema)->State_Schema:
    title = State['title']
    outline = State['outline']
    prompt = f"genrate content for the following topic:{title} and these are the outline about it{outline}"
    res = model.invoke(prompt)
    State['content'] = res.content
    return State

def evaluation(State:State_Schema)->State_Schema:
    title = State['title']
    content = State['content']
    prompt = f"evalute this {content} for the title {title} and give a score out of 10 to it ."
    res = model.invoke(prompt)
    State['score'] = res.content
    return State
    

```

```python
# add nodes
graph.add_node('genrate_outlin' , gen_outline)
graph.add_node('genrate_blogg' , gen_blog)
graph.add_node('evaluationn' , evaluation)

# add edges
graph.add_edge(START , 'genrate_outlin' )
graph.add_edge('genrate_outlin' ,  'genrate_blogg')
graph.add_edge('genrate_blogg' ,  'evaluationn')
graph.add_edge('evaluationn' , END)

# compile graph 

workflow = graph.compile()
```

```python
res = workflow.invoke({'title':"will ai replace the traditional software devlopment?"})
```
