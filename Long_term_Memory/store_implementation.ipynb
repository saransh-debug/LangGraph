```python
from langgraph.store.memory import InMemoryStore
```

```python
store = InMemoryStore()
```

```python
namespace = ("user","u1")
namespace2 = ("user","u2")
```

```python
store.put(namespace,"1",{"data":"user likes pizza."})
store.put(namespace,"2",{"data":"user like cheesy pasta."})
store.put(namespace2,"1",{"data":"user like pista."})
store.put(namespace2,"2",{"data":"user likes sphagettii."})


```

```python
store.get(namespace,"1")
store.get(namespace2,"2")
```

```python
items = store.search(namespace2)

for item in items:
    print(item.value)
```

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings

embeddings_model = GoogleGenerativeAIEmbeddings( model = "models/gemini-embedding-001")

store = InMemoryStore(index={'embed':embeddings_model})
```

```python
store.put(namespace, "user_fact_1", {"data": "The user enjoys trying new pizza toppings."})
store.put(namespace, "user_fact_2", {"data": "The user prefers working in the morning."})
store.put(namespace, "user_fact_3", {"data": "The user keeps a list of interesting ideas."})
store.put(namespace, "user_fact_4", {"data": "The user likes learning through hands-on examples."})
store.put(namespace, "user_fact_5", {"data": "The user enjoys listening to music while working."})
store.put(namespace, "user_fact_6", {"data": "The user is curious about how AI systems work."})
store.put(namespace, "user_fact_7", {"data": "The user prefers organized notes."})
store.put(namespace, "user_fact_8", {"data": "The user likes exploring new programming tools."})
store.put(namespace, "user_fact_9", {"data": "The user enjoys solving logic puzzles."})
store.put(namespace, "user_fact_10", {"data": "The user appreciates concise explanations."})
```

```python
items = store.search(namespace,query="what is the preference of the user ?",limit=3)

for item in items:
    print(item.value)
```
